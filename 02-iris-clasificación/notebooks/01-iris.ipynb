## Entrenamiento del modelo

Scaler sirve para poner todas las variables numéricas en una misma escala o rango comparable.

Luego tenemos que buscar la mejor configuración para nuestro modelo de regresión logística.

Está , lo que hará será **calcular una probabilidad entre 0 y 1**
Como el dataset Iris tiene 3 especies. nos dará la probabilidad para cada.
Usando "penalty" que es un tipo de regularización, sirve para evitar que el modelo memorice demasiado los datos (overfitting)
Usando "l2" penaliza los coeficientes grandes suavizándolos (Ridge)
C es la fuera de regularización: Cuanto mas pequeño es C, más fuerte es la penalización (modelo mas simple).
Cuanto más grande es C, menos se penaliza (modelo más ajustado a los datos).


```python
scaler = StandardScaler() #Definimos el scaler
X_train_scaled = scaler.fit_transform(X_train) #Tanto para las filas del modelo
X_test_scaled = scaler.transform(X_test) #Como para las filas del test
```

```python
X_train_scaled
```

```python
#Creamos el modelo de regresion logística, con 1000 nos aseguramos que aprende sin problemas
log_reg = LogisticRegression(max_iter=1000)

#Opciones para ver cual funciona mejor
param_grid_log_reg = {
    'penalty': ['none', 'l2'],
    'C': np.logspace(-4,4, 10)
}

#Elige combinaciones al azar para encontrar una óptima rápidamente
#Dividimos el entrenamiento en 5 partes. Entrena con 4 y valida con la restante, repitiendo eso 5 veces.Asi asegura un resultado real.
#Con n_jobs le esta diciendo a python que use todos los núcles del procesador en paralelo para que la búsqueda termine mucho más rápido
grid_search_log_reg = RandomizedSearchCV(log_reg, param_grid_log_reg, cv=5, n_jobs=-1)

#Ejecutar la búsqueda y entrenar
grid_search_log_reg.fit(X_train_scaled, y_train)

#Guardamos el modelo ganador
best_log_reg = grid_search_log_reg.best_estimator_
```
