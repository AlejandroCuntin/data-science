## Entrenamiento del modelo

Scaler sirve para poner todas las variables numéricas en una misma escala o rango comparable.

Modelo de regresión logística:
Tenemos que buscar la mejor configuración:

Está , lo que hará será **calcular una probabilidad entre 0 y 1**
Como el dataset Iris tiene 3 especies. nos dará la probabilidad para cada.
`penalty`: un tipo de regularización, sirve para evitar que el modelo memorice demasiado los datos (overfitting)
`l2`: penaliza los coeficientes grandes suavizándolos (Ridge)
`C` es la fuera de regularización
- Cuanto mas pequeño es C, más fuerte es la penalización (modelo mas simple).
- Cuanto más grande es C, menos se penaliza (modelo más ajustado a los datos).

Modelo de Random Forest:

Random Forest consiste en entrenar cientos de árboles de decisión donde cada uno ve flores y medidas distintas para que se copien entre sí. Al final, clasificamos la flor por mayoría de votos de todos los árboles, logrando un modelo mucho más robusto que no se confunde con flores atípicas.

**Muestreo aleatorio de flores (Bootstrapping)**: Cada árbol se entrena con un grupo de flores ligeramente distinto seleccionado al azar.
**Medidas aleatorias en cada corte**: En cada pregunta, el árbol no puede elegir entre las 4 medidas, se le obliga a elegir entre solo 2 al azar. Esto fuerza a que los árboles sean diferentes entre sí y busquen patrones distintos.
**Decisión por consenso**: CUando llega una flor nueva del conjunto de test:
- Los 100 árboles examinan sus medidas y votan.
- Si 92 árboles dicen que es *Versicolor* y 8 dicen que es *Virginica*, el modelo predice **Versicolor*. Al promediar tantos árboles distintos, los errores individuales se anulan.

`n_estimators`: Cuántos árboles participan en la decisión.
`max_depth=5`:(el ganador). Le prohibimos a los árboles.

````python
scaler = StandardScaler() #Definimos el scaler
X_train_scaled = scaler.fit_transform(X_train) #Tanto para las filas del modelo
X_test_scaled = scaler.transform(X_test) #Como para las filas del test
````

````python
X_train_scaled
````

````python
#Creamos el modelo de regresion logística, con 1000 nos aseguramos que aprende sin problemas
log_reg = LogisticRegression(max_iter=1000)



#Opciones de hiperparámetros (quitamos 'none' para que no dé error)
param_grid_log_reg = {
    'C': np.logspace(-4,4, 10)
}

#Busqueda aleatoria con validación cruzada(cv=5)

#Elige combinaciones al azar para encontrar una óptima rápidamente
#Dividimos el entrenamiento en 5 partes. Entrena con 4 y valida con la restante, repitiendo eso 5 veces.Asi asegura un resultado real.
#Con n_jobs le esta diciendo a python que use todos los núcles del procesador en paralelo para que la búsqueda termine mucho más rápido
grid_search_log_reg = RandomizedSearchCV(log_reg, param_grid_log_reg, cv=5, n_jobs=-1, random_state=42)

#Entranar y guardar el modelo ganador
grid_search_log_reg.fit(X_train_scaled, y_train)
best_log_reg = grid_search_log_reg.best_estimator_

#Ver qué hiperparámetro ganó y su nota media en validación cruzada (Train)
print("Mejores hiperparámetros (LogReg):", grid_search_log_reg.best_params_)
print(f"CV Accuracy en Train (LogReg):{grid_search_log_reg.best_score_: 4f}")
````

````python
best_log_reg
````

````python
#Creamos el RandomForest
rf = RandomForestClassifier(random_state=42)
param_grid_rf = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 5, 10, 20]
}

grid_search_rf = RandomizedSearchCV(rf, param_grid_rf, cv=5, n_jobs=-1, random_state=42)
grid_search_rf.fit(X_train_scaled, y_train)
best_rf = grid_search_rf.best_estimator_

#Ver qué combinacion de ároles y profundidad ganó y su nota media en Train
print("Mejores hiperparámetros (Random Forest):", grid_search_rf.best_params_)
print(f"CV Accuracy en Train (Random Forest): {grid_search_rf.best_score_: 4f}")
````

### Explicación de los resultados:

`best_params`: Te dice exactamente qué valor de `C` (en Regresión Logística) 0 qué `n_estimators` y `max_depth` (en Random Forest) fueron los ganadores.

`best_score_`: Te da la precisión media de las 5 particiones de validación cruzada durante el entrenamiento.

`random_state=42`: Hace que el sorte aleatorio dé siempre el mismo resultado cada vez que ejecutamos la celda.

## Arbol de decision simple

Es un algoritmo que aprende a clasificar las flores creando un árbol de reglas del tipo `Sí/No` basadas enc ortes numéricos de las medidas:

- Primero buscar qué medida separa mejor a las flores de un solo tajo (por ejemplo el largo del petalo).Sabemos que si el largo del petalo es menor que 2.45cm si o si se sabe seguro que es **Setosa**.
- Si es No, baja a la siguiente rama y hace otra pregunta para separar *Versicolor de Virginica** (por ejemplo si el acho del pétalo es menor que 1.75cm)

Sirve de línea base para justificar Random Forest. Un solo árbol tiene un gran defecto: si le dejas hacer demasiadas preguntas, **memoriza las flores de entrenamiento (overfitting)** y es muy inestable. Al entrenarlo primero, demostramos por qué luego necesitamos un Random Forest (juntar 200 árboles distintos votando para corregir los errores de un árbol solitario)

`max_depth`: (Profundidad máxima): El límite de pisos de preguntas que puede hacer hacia abajo para que no memorice
`min_samples_split`: El mínimo de flores que tiene que haber en un grupo para permitirle dividirlo en otra rama

¿Si ya tenemos otros modelos para que queremos este de aqui? 

Porque estamos haciendo **un "casting" (una competición) de algoritmos**: En cualquier proyecto de Machine Learning no sabemos de entrada qué algoritmo va a funcionar mejor con los datos que tenemos. Así que ponemos a competir a **4 candidatos** que piensan de formas distintas **(Formulas de probabilidad vs 1 Árbol de preguntas vs 200 Árboles votando vs Fronteras geométricas) y nos quedamos con el que saque mejor nota

````python
from sklearn.tree import DecisionTreeClassifier

#Creamos un solo árbol de decisión
tree = DecisionTreeClassifier(random_state=42)

#Probamos distitntas profundidades máximas de preguntas
param_grid_tree = {
    'max_depth': [None, 2, 3, 5, 10],
    'min_samples_split': [2, 5, 10]
}

#Buscamos la mejor configuración con validación cruzada (cv=5)
grid_search_tree = RandomizedSearchCV(
    tree, param_grid_tree, cv=5, n_jobs=-1, random_state=42
)

grid_search_tree.fit(X_train_scaled, y_train)
best_tree = grid_search_tree.best_estimator_

print("Mejores hiperparámetros (Árbol de Decisión):", grid_search_tree.best_params_)
print(f"CV Accuracy en Train (Árbol de Decisión): {grid_search_tree.best_score_: 4f}")
````
